# Data Structures and Collections — Week 2

This notebook introduces the containers that help Python programs organize information. We will begin with lists, then move to tuples, sets, and dictionaries.


## Why collections matter

Real programs rarely work with just one value. A school app may track many students. A weather app may store many temperatures. Collections help us group related data cleanly.


## Lists: ordered and changeable

A list keeps items in order, and we can edit it as the program runs.


```python
favorite_topics = ["variables", "strings", "lists"]
print(favorite_topics)
print(type(favorite_topics))
print("Number of topics:", len(favorite_topics))
```

```python
study_plan = ["review notes", "solve 3 problems", "take a short break", "write reflections"]

print("First task:", study_plan[0])
print("Last task:", study_plan[-1])
```

### Updating a list

Because lists are mutable, we can add, remove, and replace items.


```python
study_plan.append("practice with a partner")
study_plan[1] = "solve 5 problems"

print(study_plan)
```

## Slicing

Slicing lets us look at part of a list without changing the original list.


```python
weekly_scores = [72, 78, 81, 85, 90, 88, 93]
print("First three scores:", weekly_scores[:3])
print("Middle scores:", weekly_scores[2:5])
print("Every other score:", weekly_scores[::2])
```

## Tuples: ordered and stable

A tuple is useful when the values belong together and should not change accidentally.


```python
rgb_color = (32, 145, 210)
print(rgb_color)
print(type(rgb_color))
```

## Sets: unique items only

Sets automatically remove duplicates, which makes them useful for checking distinct values.


```python
skills_practiced = {"loops", "functions", "loops", "strings"}
print(skills_practiced)
print("Unique skills practiced:", len(skills_practiced))
```

## Dictionaries: key–value pairs

A dictionary stores labeled information. This is one of the most useful structures in Python.


```python
student_profile = {
    "name": "Avery",
    "level": "beginner",
    "completed_lessons": 6,
    "favorite_topic": "strings",
}

print(student_profile)
print(student_profile["name"])
print(student_profile["completed_lessons"])
```

```python
student_profile["completed_lessons"] += 1
student_profile["goal"] = "build a small portfolio project"

for key, value in student_profile.items():
    print(f"{key}: {value}")
```

## Choosing the right collection

- use a **list** when order matters and changes are expected
- use a **tuple** for a fixed group of values
- use a **set** when uniqueness matters
- use a **dictionary** when labels matter
