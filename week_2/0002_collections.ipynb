# Data Structures and Collections — Week 2

This notebook introduces the containers that help Python programs organize related pieces of data.


## Learning goals

You will learn to:

- choose a collection type based on the job
- reuse string formatting, variables, loops, and indexing from earlier lessons


## Why collections matter

Real programs almost always work with groups of values. A class roster, a shopping cart, and a list of scores are all examples of grouped data.


## Lists: ordered and changeable


```python
favorite_topics = ["variables", "strings", "lists"]
print(favorite_topics)
print(type(favorite_topics))
print("Number of topics:", len(favorite_topics))
```

```python
study_plan = ["review notes", "solve 3 problems", "take a short break", "write reflections"]

print("First task:", study_plan[0])
print("Last task:", study_plan[-1])
```

## Updating a list

Lists are mutable, so we can change them over time.


```python
study_plan.append("practice with a partner")
study_plan[1] = "solve 5 problems"

print(study_plan)
```

## Slicing a list

Slicing lets us inspect part of the collection without changing it.


```python
weekly_scores = [72, 78, 81, 85, 90, 88, 93]
print("First three scores:", weekly_scores[:3])
print("Middle scores:", weekly_scores[2:5])
print("Every other score:", weekly_scores[::2])
```

## Reusing loops from Week 1 and Week 2

Collections become much more useful when we iterate over them.


```python
for task in study_plan:
    print(f"Planned task: {task}")
```

## Tuples: ordered and stable

Use a tuple when values belong together and should stay fixed.


```python
rgb_color = (32, 145, 210)
print(rgb_color)
print(type(rgb_color))
print(f"Red channel: {rgb_color[0]}")
```

## Sets: unique values only

Sets remove duplicates automatically.


```python
skills_practiced = {"loops", "functions", "loops", "strings"}
print(skills_practiced)
print("Unique skills practiced:", len(skills_practiced))
```

## Dictionaries: key–value pairs

Dictionaries are one of the most important structures in Python.


```python
student_profile = {
    "name": "Avery",
    "level": "beginner",
    "completed_lessons": 6,
    "favorite_topic": "strings",
}

print(student_profile)
print(student_profile["name"])
print(student_profile["completed_lessons"])
```

```python
student_profile["completed_lessons"] += 1
student_profile["goal"] = "build a small portfolio project"

for key, value in student_profile.items():
    print(f"{key}: {value}")
```

## Comparing collection choices

Ask these questions:
- Do I need order?
- Do I need labels?
- Do values have to stay unique?
- Will the collection change later?


```python
shopping_list = ["notebook", "pencil", "eraser"]
item_prices = {"notebook": 3.5, "pencil": 1.0, "eraser": 0.75}

for item in shopping_list:
    print(f"{item}: ${item_prices[item]:.2f}")
```
