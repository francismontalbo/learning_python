# Unit Testing Basics

Testing helps us confirm that our code behaves the way we expect.


## Learning goals

You will learn to:
- write simple assertions
- test several input cases systematically
- connect testing to functions from earlier notebooks


```python
def add_numbers(a, b):
    """Return the sum of two numbers."""
    return a + b

print(add_numbers.__doc__)
print(add_numbers(3, 3))
```

## A first manual test


```python
def test_add_numbers():
    assert add_numbers(2, 3) == 5
    assert add_numbers(-1, 1) == 0
    assert add_numbers(0, 0) == 0


test_add_numbers()
print("All manual tests passed.")
```

## Several test cases in a loop


```python
test_cases = [
    (2, 3, 5),
    (10, -2, 8),
    (3.5, 4.5, 8.0),
]

for left, right, expected in test_cases:
    result = add_numbers(left, right)
    assert result == expected, f"Expected {expected}, got {result}"

print("Batch test run completed successfully.")
```

## Testing another function from earlier topics


```python
def is_even(number):
    return number % 2 == 0

assert is_even(0) is True
assert is_even(8) is True
assert is_even(7) is False
print("is_even tests passed.")
```

## Why testing matters

Tests turn expectations into executable checks. That is one of the habits that separates guesswork from engineering.
