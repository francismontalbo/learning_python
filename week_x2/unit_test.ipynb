# Unit Testing Basics

Testing helps us confirm that our code behaves the way we expect. It also gives us confidence to improve code later without breaking old behavior.


```python
def add_numbers(a, b):
    """Return the sum of two numbers."""
    return a + b

print(add_numbers.__doc__)
print(add_numbers(3, 3))
```

## A first manual test

An `assert` statement checks whether a condition is true. If the condition is false, Python raises an error.


```python
def test_add_numbers():
    assert add_numbers(2, 3) == 5
    assert add_numbers(-1, 1) == 0
    assert add_numbers(0, 0) == 0


test_add_numbers()
print("All manual tests passed.")
```

## Testing several cases systematically

Using a list of test cases keeps the testing code compact and readable.


```python
test_cases = [
    (2, 3, 5),
    (10, -2, 8),
    (3.5, 4.5, 8.0),
]

for left, right, expected in test_cases:
    result = add_numbers(left, right)
    assert result == expected, f"Expected {expected}, got {result}"

print("Batch test run completed successfully.")
```

## Why testing is professional

Tests turn your expectations into executable checks. That is one of the habits that separates guesswork from engineering.
