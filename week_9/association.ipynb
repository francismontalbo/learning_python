# Association

Association describes a relationship between objects that know about each other, but neither object necessarily owns the other.


```python
class Address:
    def __init__(self, street, city, postal_code):
        self.street = street
        self.city = city
        self.postal_code = postal_code

    def formatted(self):
        return f"{self.street}, {self.city}, {self.postal_code}"


class Person:
    def __init__(self, name, address):
        self.name = name
        self.address = address

    def introduce(self):
        print(f"Hello, I am {self.name}.")

    def show_address(self):
        print(self.address.formatted())
```

```python
home_address = Address("15 Garden Lane", "Maple Town", "40210")
resident = Person("Avery", home_address)

resident.introduce()
resident.show_address()
```

## Another example

A teacher may know a course, and a course may know its teacher. The relationship is meaningful, but neither object has to fully own the other.


```python
class Course:
    def __init__(self, title):
        self.title = title


class Teacher:
    def __init__(self, name, course):
        self.name = name
        self.course = course

    def teach(self):
        print(f"{self.name} teaches {self.course.title}.")

python_course = Course("Python Foundations")
teacher = Teacher("Jordan", python_course)
teacher.teach()
```
