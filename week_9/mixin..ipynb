# Mixins

A mixin is a small class that adds one focused behavior to another class.


## Learning goals

You will learn to:
- identify what a mixin should and should not do
- add narrow reusable behavior to another class
- keep inheritance design simple and intentional


```python
class TimestampMixin:
    def record_event(self, event_name):
        from datetime import datetime
        print(f"[{datetime.now().isoformat(timespec='seconds')}] {event_name}")


class Project(TimestampMixin):
    def __init__(self, title):
        self.title = title

    def start(self):
        self.record_event(f"Project '{self.title}' started")

project = Project("Weather Dashboard")
project.start()
```

## Another focused mixin


```python
class DisplayNameMixin:
    def display_name(self):
        print(f"Name: {self.name}")


class TeamMember(DisplayNameMixin):
    def __init__(self, name, role):
        self.name = name
        self.role = role

member = TeamMember("Riley", "tester")
member.display_name()
```

## Mixins should stay small

A mixin should usually contribute one narrow capability, such as logging, naming, formatting, or validation.


```python
class ActiveStateMixin:
    def is_active(self):
        return self.active


class Service(ActiveStateMixin):
    def __init__(self, name, active):
        self.name = name
        self.active = active

service = Service("Email Service", True)
print(service.is_active())
```
