# Hardware Modeling with Classes

This notebook uses simple hardware-style examples to practice object relationships. The goal is to make the object design easy to understand, not to simulate hardware perfectly.


```python
class CPU:
    def __init__(self, model, clock_speed_ghz, cores):
        self.model = model
        self.clock_speed_ghz = clock_speed_ghz
        self.cores = cores

    def summary(self):
        return f"CPU: {self.model}, {self.cores} cores at {self.clock_speed_ghz} GHz"


class RAM:
    def __init__(self, capacity_gb, speed_mhz):
        self.capacity_gb = capacity_gb
        self.speed_mhz = speed_mhz

    def summary(self):
        return f"RAM: {self.capacity_gb} GB at {self.speed_mhz} MHz"


class Storage:
    def __init__(self, kind, capacity_gb):
        self.kind = kind
        self.capacity_gb = capacity_gb

    def summary(self):
        return f"Storage: {self.capacity_gb} GB {self.kind}"
```

## Composition

In composition, the larger object creates and owns its parts directly.


```python
class Computer:
    def __init__(self):
        self.cpu = CPU("StudyChip A1", 3.2, 8)
        self.ram = RAM(16, 3200)
        self.storage = Storage("SSD", 512)

    def summary(self):
        print(self.cpu.summary())
        print(self.ram.summary())
        print(self.storage.summary())

computer = Computer()
computer.summary()
```

## Extending the design


```python
class GamingComputer(Computer):
    def __init__(self):
        super().__init__()
        self.storage = Storage("NVMe SSD", 1024)

powerful_machine = GamingComputer()
powerful_machine.summary()
```

## Key idea

The parts are created inside the larger object, so ownership is much tighter than in aggregation.
