# Hardware Modeling with Classes

This notebook uses simple hardware-style examples to practice object relationships. The goal is to make the object design easy to understand, not to simulate hardware perfectly.


```python
class CPU:
    def __init__(self, model, clock_speed_ghz, cores):
        self.model = model
        self.clock_speed_ghz = clock_speed_ghz
        self.cores = cores

    def summary(self):
        return f"CPU: {self.model}, {self.cores} cores at {self.clock_speed_ghz} GHz"


class RAM:
    def __init__(self, capacity_gb, speed_mhz):
        self.capacity_gb = capacity_gb
        self.speed_mhz = speed_mhz

    def summary(self):
        return f"RAM: {self.capacity_gb} GB at {self.speed_mhz} MHz"


class Storage:
    def __init__(self, kind, capacity_gb):
        self.kind = kind
        self.capacity_gb = capacity_gb

    def summary(self):
        return f"Storage: {self.capacity_gb} GB {self.kind}"
```

## Aggregation

In aggregation, a larger object uses smaller objects that may already exist on their own.


```python
class Computer:
    def __init__(self, cpu, ram, storage):
        self.cpu = cpu
        self.ram = ram
        self.storage = storage

    def summary(self):
        print(self.cpu.summary())
        print(self.ram.summary())
        print(self.storage.summary())

shared_cpu = CPU("StudyChip A1", 3.2, 8)
shared_ram = RAM(16, 3200)
shared_storage = Storage("SSD", 512)

workstation = Computer(shared_cpu, shared_ram, shared_storage)
workstation.summary()
```

## Reusing shared parts


```python
backup_machine = Computer(shared_cpu, RAM(8, 2666), shared_storage)
backup_machine.summary()
```

## Key idea

The parts are created first and then supplied to the larger object. That is the core of aggregation.
