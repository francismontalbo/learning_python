# Recursion

Recursion happens when a function solves a problem by calling itself on a smaller version of the same problem.


## Learning goals

You will learn to:
- identify a base case and a recursive step
- trace recursive calls carefully
- compare recursion with loops and collections from earlier notebooks


## The two required parts

Every recursive function needs:
1. a **base case** that stops the recursion
2. a **recursive step** that moves toward the base case


```python
def factorial(number):
    if number == 0:
        return 1
    return number * factorial(number - 1)

print(factorial(5))
```

## Tracing the flow

`factorial(5)` becomes `5 * factorial(4)`, then `4 * factorial(3)`, and continues until it reaches `factorial(0)`.


```python
def countdown(number):
    if number == 0:
        print("Lift off!")
        return
    print(number)
    countdown(number - 1)

countdown(5)
```

## Recursion with nested collections

This example revisits lists from earlier notebooks.


```python
def sum_nested_numbers(data):
    if isinstance(data, list):
        total = 0
        for item in data:
            total += sum_nested_numbers(item)
        return total
    return data

numbers = [1, [2, 3], [4, [5, 6]]]
print(sum_nested_numbers(numbers))
```

## Another recursive example


```python
def sum_to_n(number):
    if number == 1:
        return 1
    return number + sum_to_n(number - 1)

print(sum_to_n(5))
```

## When recursion is a good fit

Recursion is useful for nested structures, tree-like data, and problems that shrink naturally.
