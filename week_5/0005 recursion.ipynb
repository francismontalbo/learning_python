# Recursion

Recursion happens when a function solves a problem by calling itself on a smaller version of the same problem. It can feel magical at first, so we will go carefully.


## Two essential parts

A recursive function needs:
1. a **base case** that stops the recursion
2. a **recursive step** that moves toward the base case


```python
def factorial(number):
    if number == 0:
        return 1
    return number * factorial(number - 1)

print(factorial(5))
```

## Reading the flow

`factorial(5)` becomes `5 * factorial(4)`, then `4 * factorial(3)`, and so on until the function reaches `factorial(0)`.


```python
def sum_nested_numbers(data):
    if isinstance(data, list):
        total = 0
        for item in data:
            total += sum_nested_numbers(item)
        return total
    return data

numbers = [1, [2, 3], [4, [5, 6]]]
print(sum_nested_numbers(numbers))
```

## When recursion is useful

Recursion works well for tree-like or nested structures, and for problems that naturally break into smaller versions of themselves.


```python
def countdown(number):
    if number == 0:
        print("Lift off!")
        return
    print(number)
    countdown(number - 1)

countdown(5)
```
