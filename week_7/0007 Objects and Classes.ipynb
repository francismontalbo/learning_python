# Objects and Classes

Object-oriented programming helps us model real things in code. A class is a blueprint, and an object is a concrete instance created from that blueprint.


## Learning goals

You will learn to:
- define classes with attributes and methods
- create multiple objects from one class
- connect classes to earlier topics such as strings, functions, and dictionaries


## Defining a class


```python
class Student:
    """Represent a student in a Python course."""

    school_name = "Bright Path Academy"

    def __init__(self, name, level):
        self.name = name
        self.level = level

    def introduce(self):
        print(f"Hello, I am {self.name} and I am a {self.level} learner.")
```

```python
print(Student.__doc__)
print(Student.school_name)
```

## Creating objects


```python
student_one = Student("Avery", "beginner")
student_two = Student("Jordan", "intermediate")

student_one.introduce()
student_two.introduce()
```

## Updating object data


```python
student_one.level = "confident beginner"
student_one.introduce()
```

## Methods can use earlier Python skills


```python
class ProgressTracker:
    def __init__(self, learner_name, completed_topics):
        self.learner_name = learner_name
        self.completed_topics = completed_topics

    def summary(self):
        print(f"{self.learner_name} completed {len(self.completed_topics)} topics.")
        for topic in self.completed_topics:
            print(f"- {topic}")

tracker = ProgressTracker("Avery", ["variables", "lists", "loops"])
tracker.summary()
```

## Another realistic class example


```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if amount > self.balance:
            return "Insufficient funds"
        self.balance -= amount
        return self.balance

account = BankAccount("Avery", 100)
print(account.deposit(25))
print(account.withdraw(80))
print(account.withdraw(100))
```

## Objects can also return structured data


```python
class Course:
    def __init__(self, title, lesson_count):
        self.title = title
        self.lesson_count = lesson_count

    def as_dictionary(self):
        return {"title": self.title, "lesson_count": self.lesson_count}

course = Course("Python Foundations", 8)
print(course.as_dictionary())
```
