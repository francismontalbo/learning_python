# Objects and Classes

Object-oriented programming helps us model real things in code. A class is a blueprint, and an object is a concrete instance created from that blueprint.


## Defining a class

We begin with a small class so that the structure is easy to see.


```python
class Student:
    """Represent a student in a Python course."""

    school_name = "Bright Path Academy"

    def __init__(self, name, level):
        self.name = name
        self.level = level

    def introduce(self):
        print(f"Hello, I am {self.name} and I am a {self.level} learner.")
```

```python
print(Student.__doc__)
print(Student.school_name)
```

## Creating objects

Each object stores its own data in attributes such as `name` and `level`.


```python
student_one = Student("Avery", "beginner")
student_two = Student("Jordan", "intermediate")

student_one.introduce()
student_two.introduce()
```

## Updating object data

Objects can change over time, just like real records in an application.


```python
student_one.level = "confident beginner"
student_one.introduce()
```

## Adding useful methods

Methods should represent meaningful actions related to the object.


```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if amount > self.balance:
            return "Insufficient funds"
        self.balance -= amount
        return self.balance

account = BankAccount("Avery", 100)
print(account.deposit(25))
print(account.withdraw(80))
print(account.withdraw(100))
```
