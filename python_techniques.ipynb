## List comprehensions

A list comprehension builds a new list in one readable expression.


```python
numbers = list(range(1, 11))
even_squares = [number ** 2 for number in numbers if number % 2 == 0]
print(even_squares)
```

```python
fruit_names = ["apple", "banana", "cherry", "fig"]
name_lengths = [len(name) for name in fruit_names]
print(name_lengths)
```

## Flattening a nested list

A nested comprehension can gather items from smaller lists into one larger list.


```python
study_groups = [["Avery", "Jordan"], ["Riley"], ["Sam", "Taylor"]]
all_students = [student for group in study_groups for student in group]
print(all_students)
```

## A reusable helper function

This prime-checking example is longer than the earlier cells, but it demonstrates how a carefully written function can package more advanced logic.


```python
def is_prime(number):
    if number <= 1:
        return False
    if number <= 3:
        return True
    if number % 2 == 0 or number % 3 == 0:
        return False

    candidate = 5
    while candidate * candidate <= number:
        if number % candidate == 0 or number % (candidate + 2) == 0:
            return False
        candidate += 6
    return True

prime_numbers = [number for number in range(1, 50) if is_prime(number)]
print(prime_numbers)
```
