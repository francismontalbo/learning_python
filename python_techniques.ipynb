## List comprehensions


```python
numbers = list(range(1, 11))
even_squares = [number ** 2 for number in numbers if number % 2 == 0]
print(even_squares)
```

## String-based comprehensions


```python
fruit_names = ["apple", "banana", "cherry", "fig"]
name_lengths = [len(name) for name in fruit_names]
print(name_lengths)
```

## Filtering with a condition


```python
scores = [72, 85, 91, 68, 95]
strong_scores = [score for score in scores if score >= 80]
print(strong_scores)
```

## Flattening a nested list


```python
study_groups = [["Avery", "Jordan"], ["Riley"], ["Sam", "Taylor"]]
all_students = [student for group in study_groups for student in group]
print(all_students)
```

## A helper function inside a comprehension


```python
def is_prime(number):
    if number <= 1:
        return False
    if number <= 3:
        return True
    if number % 2 == 0 or number % 3 == 0:
        return False

    candidate = 5
    while candidate * candidate <= number:
        if number % candidate == 0 or number % (candidate + 2) == 0:
            return False
        candidate += 6
    return True

prime_numbers = [number for number in range(1, 50) if is_prime(number)]
print(prime_numbers)
```

## Dictionary comprehension


```python
score_labels = {score: ("pass" if score >= 75 else "review") for score in scores}
print(score_labels)
```
