# Exceptions and Error Handling

Errors are part of programming. Strong programmers do not panic when code fails; they inspect the problem, handle it carefully, and improve the program.


## Learning goals

You will learn to:
- use `try` and `except`
- catch specific exceptions
- understand `else` and `finally`
- connect error handling to functions and user-facing messages


## A simple `try` / `except`


```python
try:
    value = int("42")
    print(value)
except ValueError:
    print("That value could not be converted to an integer.")
```

```python
try:
    value = int("not-a-number")
    print(value)
except ValueError as error:
    print("Conversion failed:", error)
```

## Handling errors inside a function


```python
def parse_age(text):
    try:
        age = int(text)
        return age
    except ValueError as error:
        raise RuntimeError("Age must be written as a whole number.") from error

try:
    parse_age("twelve")
except RuntimeError as error:
    print("Handled error:", error)
    print("Original cause:", error.__cause__)
```

## Reusing conditions with exceptions


```python
def safe_divide(numerator, denominator):
    try:
        result = numerator / denominator
    except ZeroDivisionError:
        return "Cannot divide by zero."
    else:
        return f"Result: {result}"

print(safe_divide(10, 2))
print(safe_divide(10, 0))
```

## `else` and `finally`


```python
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Cannot divide by zero.")
else:
    print("Division worked:", result)
finally:
    print("This cleanup step always runs.")
```

## Why specificity matters

Catch the narrowest exception that matches the problem you expect.


```python
data = ["5", "12", "oops", "19"]
for item in data:
    try:
        print(int(item) * 2)
    except ValueError:
        print(f"Skipped invalid number: {item}")
```
