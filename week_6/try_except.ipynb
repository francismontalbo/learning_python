# Exceptions and Error Handling

Errors are part of programming. Strong programmers do not panic when code fails—they inspect the problem, handle it carefully, and improve the program.


## A simple `try` / `except`

Use `try` to run code that might fail, and `except` to respond gracefully.


```python
try:
    value = int("42")
    print(value)
except ValueError:
    print("That value could not be converted to an integer.")
```

```python
try:
    value = int("not-a-number")
    print(value)
except ValueError as error:
    print("Conversion failed:", error)
```

## Why this matters

Without error handling, one bad input can crash a program. With good error handling, you can guide the user and continue responsibly.


```python
def parse_age(text):
    try:
        age = int(text)
        return age
    except ValueError as error:
        raise RuntimeError("Age must be written as a whole number.") from error

try:
    parse_age("twelve")
except RuntimeError as error:
    print("Handled error:", error)
    print("Original cause:", error.__cause__)
```

## `else` and `finally`

- `else` runs when no exception happens
- `finally` runs whether an exception happens or not


```python
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Cannot divide by zero.")
else:
    print("Division worked:", result)
finally:
    print("This cleanup step always runs.")
```

## Professional habit

Catch the specific exception you expect. Avoid hiding unexpected problems with very broad exception handling.
