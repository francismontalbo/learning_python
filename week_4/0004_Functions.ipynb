# Functions — Week 4

Functions help us organize code into reusable pieces. They make programs easier to read, test, and improve.


## Why functions matter

When the same pattern appears more than once, a function lets us give that behavior a clear name.


```python
def add_numbers(first_number, second_number):
    """Return the sum of two numbers."""
    return first_number + second_number

print(add_numbers(4, 3))
```

## Parameters and arguments

Parameters are the names inside the function definition. Arguments are the actual values you pass in.


```python
def describe_course(course_name, lesson_count):
    return f"{course_name} includes {lesson_count} lessons."

print(describe_course("Python Foundations", 8))
```

## Return values

A strong beginner habit is to return data from a function instead of printing too early. This gives the caller more control.


```python
def calculate_area(width, height):
    return width * height

room_area = calculate_area(5, 4)
print(f"Area: {room_area}")
```

## Default arguments

Default values make a function more flexible without making it harder to use.


```python
def greet_student(name, greeting="Welcome"):
    return f"{greeting}, {name}!"

print(greet_student("Avery"))
print(greet_student("Jordan", greeting="Great to see you"))
```

## Variable-length arguments with `*args`

Use `*args` when a function should accept many positional values.


```python
def total_points(*scores):
    return sum(scores)

print(total_points(5, 10, 15))
print(total_points(7, 7, 7, 7))
```

## Small, realistic example

This calculator is intentionally limited so that the logic stays easy to understand.


```python
def apply_discount(price, percent_discount):
    discount_amount = price * (percent_discount / 100)
    return price - discount_amount

final_price = apply_discount(120, 15)
print(f"Discounted price: ${final_price:.2f}")
```
