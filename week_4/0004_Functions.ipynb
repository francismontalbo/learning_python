# Functions — Week 4

Functions help us organize code into reusable, meaningful pieces.


## Learning goals

You will learn to:
- define and call functions
- work with parameters and return values
- use default arguments and `*args`
- combine functions with loops, conditions, and collections from earlier notebooks


## A simple function


```python
def add_numbers(first_number, second_number):
    """Return the sum of two numbers."""
    return first_number + second_number

print(add_numbers(4, 3))
```

## Parameters and arguments


```python
def describe_course(course_name, lesson_count):
    return f"{course_name} includes {lesson_count} lessons."

print(describe_course("Python Foundations", 8))
```

## Returning values

Returning data gives the caller more freedom than printing too early.


```python
def calculate_area(width, height):
    return width * height

room_area = calculate_area(5, 4)
print(f"Area: {room_area}")
```

## Functions with conditions


```python
def check_passing_score(score):
    if score >= 75:
        return "pass"
    return "review"

print(check_passing_score(82))
print(check_passing_score(60))
```

## Default arguments


```python
def greet_student(name, greeting="Welcome"):
    return f"{greeting}, {name}!"

print(greet_student("Avery"))
print(greet_student("Jordan", greeting="Great to see you"))
```

## Variable-length arguments with `*args`


```python
def total_points(*scores):
    return sum(scores)

print(total_points(5, 10, 15))
print(total_points(7, 7, 7, 7))
```

## Functions with loops


```python
def print_tasks(tasks):
    for task in tasks:
        print(f"Task: {task}")

print_tasks(["read notes", "solve exercise", "rest"])
```

## A realistic example


```python
def apply_discount(price, percent_discount):
    discount_amount = price * (percent_discount / 100)
    return price - discount_amount

final_price = apply_discount(120, 15)
print(f"Discounted price: ${final_price:.2f}")
```

## Functions returning collections


```python
def build_progress_report(name, completed_lessons):
    return {
        "name": name,
        "completed_lessons": completed_lessons,
        "status": "on track" if completed_lessons >= 5 else "keep practicing",
    }

report = build_progress_report("Avery", 6)
print(report)
```
