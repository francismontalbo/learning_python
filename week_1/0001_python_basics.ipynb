## Variables store information

A variable is a name that points to a value. Choose names that describe the meaning of the data.


```python
student_name = "Avery"
study_hours = 2
confidence_score = 8.5
is_ready_to_practice = True

print(student_name)
print(type(study_hours))
print(type(confidence_score))
print(type(is_ready_to_practice))
```

## Basic data types

For now, focus on four common types:
- `int` for whole numbers
- `float` for numbers with decimals
- `str` for text
- `bool` for `True` and `False`


```python
whole_number = 12
decimal_number = 12.75
message = "Python is readable."
finished = False

print(whole_number, type(whole_number))
print(decimal_number, type(decimal_number))
print(message, type(message))
print(finished, type(finished))
```

## Arithmetic operators

Python can add, subtract, multiply, divide, and more. Notice how clear variable names make the math easier to follow.


```python
weekly_goal = 20
completed_lessons = 7
remaining_lessons = weekly_goal - completed_lessons
practice_sets = 3 * 4
average_score = 17 / 4
whole_groups = 17 // 4
leftover_items = 17 % 4
power_result = 2 ** 5

print("Remaining lessons:", remaining_lessons)
print("Practice sets:", practice_sets)
print("Average score:", average_score)
print("Whole groups:", whole_groups)
print("Leftover items:", leftover_items)
print("2 to the power of 5:", power_result)
```

## Strings and f-strings

F-strings let us place variables directly inside text. They are one of the clearest ways to build readable output.


```python
learner = "Avery"
language = "Python"
lesson_number = 1

summary = f"{learner} is studying {language} in lesson {lesson_number}."
print(summary)
```

## Small professional habits

When you write code, try to:
1. use descriptive variable names
2. keep examples short
3. test one idea at a time
4. read your output carefully


```python
price_per_book = 14.99
number_of_books = 3
total_cost = price_per_book * number_of_books

print(f"Buying {number_of_books} books costs ${total_cost:.2f}.")
```
