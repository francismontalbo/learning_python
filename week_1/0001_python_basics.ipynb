## 1. Your first Python output

The `print()` function sends text or values to the screen.


```python
print("Hello, Python learner!")
print("This is the beginning of your Python journey.")
```

## 2. Comments

Comments help humans understand code. Python ignores comments when it runs the program.


```python
# This line prints a message.
print("Comments are for people, not for Python.")

# Try uncommenting the next line.
# print("I am practicing by editing code.")
```

## 3. Variables

A variable is a name that stores a value so you can use that value again later.


```python
student_name = "Avery"
study_hours = 2
confidence_score = 8.5
is_ready_to_practice = True

print(student_name)
print(study_hours)
print(confidence_score)
print(is_ready_to_practice)
```

## 4. Data types

Four beginner-friendly types appear all the time:
- `str` for text
- `int` for whole numbers
- `float` for decimals
- `bool` for `True` and `False`


```python
print(type(student_name))
print(type(study_hours))
print(type(confidence_score))
print(type(is_ready_to_practice))
```

## 5. Whole numbers and decimal numbers


```python
whole_number = 12
decimal_number = 12.75
negative_number = -4

print(whole_number)
print(decimal_number)
print(negative_number)
```

## 6. Strings

Strings store text such as names, labels, messages, and sentences.


```python
course_title = "Python Foundations"
lesson_message = "Readable code is a professional habit."
quote_message = "Learning slowly is still learning well."

print(course_title)
print(lesson_message)
print(quote_message)
```

## 7. Arithmetic operators

Python supports addition, subtraction, multiplication, division, floor division, modulus, and exponentiation.


```python
a = 15
b = 4

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Exponent:", a ** 2)
```

## 8. Arithmetic in a realistic mini-scenario


```python
weekly_goal = 20
completed_lessons = 7
remaining_lessons = weekly_goal - completed_lessons
practice_sets = 3 * 4
average_score = 17 / 4
whole_groups = 17 // 4
leftover_items = 17 % 4
power_result = 2 ** 5

print("Remaining lessons:", remaining_lessons)
print("Practice sets:", practice_sets)
print("Average score:", average_score)
print("Whole groups:", whole_groups)
print("Leftover items:", leftover_items)
print("2 to the power of 5:", power_result)
```

## 9. Updating variables

Variables can change as a program runs.


```python
completed_lessons = completed_lessons + 3
study_hours += 1
confidence_score = confidence_score + 0.5

print("Updated completed lessons:", completed_lessons)
print("Updated study hours:", study_hours)
print("Updated confidence score:", confidence_score)
```

## 10. Assignment step by step

Many beginners become stronger when they see values change line by line.


```python
score = 10
print("Initial score:", score)

score = score + 5
print("After bonus:", score)

score = score * 2
print("After doubling:", score)
```

## 11. Casting and type conversion

Casting means converting one type into another type.


```python
whole_number_text = "25"
decimal_text = "7.5"
real_number = 19

as_integer = int(whole_number_text)
as_float = float(decimal_text)
as_string = str(real_number)

print(as_integer, type(as_integer))
print(as_float, type(as_float))
print(as_string, type(as_string))
```

## 12. Boolean conversions

Python can also convert some values to `bool`.


```python
print(bool(1))
print(bool(0))
print(bool("Python"))
print(bool(""))
```

## 13. Strings and concatenation

We can join strings together using `+`, but everything must be a string.


```python
first_name = "Avery"
last_name = "Stone"
full_name = first_name + " " + last_name

print(full_name)
```

## 14. Mixing text and numbers

If you want to join a number to text using `+`, cast the number first.


```python
age = 14
print("Age:", age)
print("I am " + str(age) + " years old.")
```

## 15. f-strings

F-strings are one of the best beginner tools in modern Python.


```python
summary = f"{student_name} studied for {study_hours} hours and completed {completed_lessons} lessons."
print(summary)
```

## 16. Other string formatting styles

You may still see older formatting approaches in real code, so it helps to recognize them.


```python
job = "Engineer"
name = "Morgan"
age = 28

print("Hello, {}!".format(name))
print("{} is a {}.".format(name, job))
print("%s is %d years old." % (name, age))
```

## 17. Quotation marks in strings

Python supports both single quotes and double quotes.


```python
message_one = "Python is fun"
message_two = 'You can also use single quotes'
message_three = "It's okay to include an apostrophe like this."

print(message_one)
print(message_two)
print(message_three)
```

## 18. Repeating strings

The `*` operator can repeat a string.


```python
print("-" * 20)
print("Practice! " * 3)
print("=" * 20)
```

## 19. String indexing

Strings are ordered, so you can access characters by position.


```python
language = "Python"
print(language[0])
print(language[1])
print(language[-1])
```

## 20. String slicing

Slicing lets you take part of a string.


```python
word = "developer"
print(word[:4])
print(word[4:])
print(word[1:6])
```

## 21. Useful string methods


```python
topic = "python basics"
print(topic.upper())
print(topic.title())
print(topic.replace("basics", "journey"))
print(len(topic))
```

## 22. Comparison operators

Comparison operators answer true-or-false questions.


```python
x = 5
y = 7

print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

## 23. Comparing real values in context


```python
apple_price = 20
orange_price = 20
iphone_cost = 60000
android_cost = 35000

print(apple_price == orange_price)
print(iphone_cost > android_cost)
print(iphone_cost < android_cost)
```

## 24. Logical operators

Logical operators combine Boolean values.


```python
p = True
v = False

print(p and v)
print(p or v)
print(not p)
print(True and True and False)
print(False or False or True)
```

## 25. Booleans in practical conditions


```python
finished_homework = True
attended_class = True

ready_for_quiz = finished_homework and attended_class
print(ready_for_quiz)
```

## 26. Simulating user input safely

Interactive `input()` examples are important, but notebooks can become awkward if they always pause for typing. So we will simulate the result of user input with strings.


```python
first_input = "12"
second_input = "8"

print(type(first_input))
print(type(second_input))
print(first_input + second_input)  # string concatenation, not number addition
```

## 27. Turning simulated input into numbers


```python
first_number = int(first_input)
second_number = int(second_input)
answer = first_number + second_number

print(type(first_number))
print(type(second_number))
print(answer)
```

## 28. Floating-point formatting

You can control how many decimal places appear in output.


```python
num1 = 7
num2 = 3
result = num1 / num2

print(result)
print(f"Formatted to 2 decimal places: {result:.2f}")
print(f"Formatted to 4 decimal places: {result:.4f}")
```

## 29. Bringing strings, numbers, and formatting together


```python
price_per_book = 14.99
number_of_books = 3
total_cost = price_per_book * number_of_books

receipt = f"Buying {number_of_books} books costs ${total_cost:.2f}."
print(receipt)
```

## 31. Procedural mini-program: study planner


```python
subject = "Python"
minutes_per_session = 30
sessions_per_week = 5
weekly_minutes = minutes_per_session * sessions_per_week

print(f"Subject: {subject}")
print(f"Minutes per session: {minutes_per_session}")
print(f"Sessions per week: {sessions_per_week}")
print(f"Total weekly minutes: {weekly_minutes}")
```

## 32. Procedural mini-program: shopping math


```python
notebook_price = 3.50
pencil_price = 1.25
eraser_price = 0.75

total = notebook_price + pencil_price + eraser_price
print(f"Total school supply cost: ${total:.2f}")
```

## 33. Procedural mini-problem: animal legs


```python
chickens = 20
rabbits = 15
total_legs = chickens * 2 + rabbits * 4

print("Total legs on the farm:", total_legs)
```

## 34. Procedural mini-problem: game coins


```python
coins = 10
earnings_per_win = 3
target = 20
games_needed = (target - coins) / earnings_per_win

print("Games needed to reach the target:", games_needed)
```
