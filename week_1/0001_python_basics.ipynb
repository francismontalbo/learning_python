## Variables store information

Variables let us name data so we can reuse it later.


```python
student_name = "Avery"
study_hours = 2
confidence_score = 8.5
is_ready_to_practice = True

print(student_name)
print(study_hours)
print(confidence_score)
print(is_ready_to_practice)
```

## Checking data types

Python can tell us what kind of value a variable currently holds.


```python
print(type(student_name))
print(type(study_hours))
print(type(confidence_score))
print(type(is_ready_to_practice))
```

## Numbers in Python

You will often work with whole numbers (`int`) and decimal numbers (`float`).


```python
whole_number = 12
decimal_number = 12.75

print(whole_number, type(whole_number))
print(decimal_number, type(decimal_number))
```

## Text in Python

Strings store text. They are used for names, labels, messages, and many kinds of program output.


```python
course_title = "Python Foundations"
lesson_message = "Readable code is a professional habit."

print(course_title)
print(lesson_message)
```

## Arithmetic operators

Let us use arithmetic to model a simple study scenario.


```python
weekly_goal = 20
completed_lessons = 7
remaining_lessons = weekly_goal - completed_lessons
practice_sets = 3 * 4
average_score = 17 / 4
whole_groups = 17 // 4
leftover_items = 17 % 4
power_result = 2 ** 5

print("Remaining lessons:", remaining_lessons)
print("Practice sets:", practice_sets)
print("Average score:", average_score)
print("Whole groups:", whole_groups)
print("Leftover items:", leftover_items)
print("2 to the power of 5:", power_result)
```

## Updating variables

Programs often change values as work gets completed.


```python
completed_lessons = completed_lessons + 3
study_hours += 1

print("Updated completed lessons:", completed_lessons)
print("Updated study hours:", study_hours)
```

## f-strings for readable output

F-strings help us mix variables and text cleanly.


```python
summary = f"{student_name} studied for {study_hours} hours and completed {completed_lessons} lessons."
print(summary)
```

## Bringing earlier topics together

Here we combine variables, arithmetic, and strings in one short example.


```python
price_per_book = 14.99
number_of_books = 3
total_cost = price_per_book * number_of_books

receipt = f"Buying {number_of_books} books costs ${total_cost:.2f}."
print(receipt)
```
