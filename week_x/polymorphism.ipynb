# Polymorphism

Polymorphism means that different objects can respond to the same method name in their own way. This makes code flexible and extensible.


```python
class EnglishGreeting:
    def say_hello(self):
        print("Hello")


class SpanishGreeting:
    def say_hello(self):
        print("Hola")


class JapaneseGreeting:
    def say_hello(self):
        print("Konnichiwa")
```

```python
def introduce(greeting_style):
    greeting_style.say_hello()

for greeter in [EnglishGreeting(), SpanishGreeting(), JapaneseGreeting()]:
    introduce(greeter)
```

## The key idea

The `introduce()` function does not care about the concrete class. It only cares that the object knows how to run `say_hello()`.


```python
class Rectangle:
    def area(self, width, height):
        return width * height


class Triangle:
    def area(self, width, height):
        return (width * height) / 2

shapes = [Rectangle(), Triangle()]
for shape in shapes:
    print(shape.area(6, 4))
```
