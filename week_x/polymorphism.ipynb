# Polymorphism

Polymorphism means different objects can respond to the same method name in their own way.


## Learning goals

You will learn to:
- use one shared interface across different classes
- write flexible functions that accept many object types
- connect this idea to methods and classes from earlier notebooks


```python
class EnglishGreeting:
    def say_hello(self):
        print("Hello")


class SpanishGreeting:
    def say_hello(self):
        print("Hola")


class JapaneseGreeting:
    def say_hello(self):
        print("Konnichiwa")
```

```python
def introduce(greeting_style):
    greeting_style.say_hello()

for greeter in [EnglishGreeting(), SpanishGreeting(), JapaneseGreeting()]:
    introduce(greeter)
```

## Another polymorphism example


```python
class Rectangle:
    def area(self, width, height):
        return width * height


class Triangle:
    def area(self, width, height):
        return (width * height) / 2

shapes = [Rectangle(), Triangle()]
for shape in shapes:
    print(shape.area(6, 4))
```

## Reusing polymorphism with loops


```python
class EmailNotifier:
    def send(self, message):
        print(f"Email: {message}")


class SMSNotifier:
    def send(self, message):
        print(f"SMS: {message}")


def broadcast(services, message):
    for service in services:
        service.send(message)

broadcast([EmailNotifier(), SMSNotifier()], "Practice session starts at 4 PM.")
```
