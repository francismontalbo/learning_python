# Encapsulation

Encapsulation means keeping data and behavior together while controlling how important values can be changed.


## Learning goals

You will learn to:
- protect internal state with controlled methods
- validate changes before storing them
- connect encapsulation to classes, conditions, and exceptions


```python
class Car:
    def __init__(self, speed, color):
        self._speed = speed
        self.color = color

    def get_speed(self):
        return self._speed

    def set_speed(self, new_speed):
        if new_speed < 0:
            raise ValueError("Speed cannot be negative.")
        self._speed = new_speed
```

```python
car = Car(80, "blue")
print(car.get_speed())
car.set_speed(95)
print(car.get_speed())
```

## Validating bad input


```python
try:
    car.set_speed(-10)
except ValueError as error:
    print(error)
```

## Another encapsulation example


```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance

    def get_balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")
        self._balance += amount

    def withdraw(self, amount):
        if amount > self._balance:
            raise ValueError("Insufficient funds.")
        self._balance -= amount

account = BankAccount("Avery", 100)
account.deposit(25)
account.withdraw(40)
print(account.get_balance())
```

## Naming convention

A single underscore, like `_balance`, is a Python signal that the attribute is meant for internal use.
