# Encapsulation

Encapsulation means keeping an object's data and behavior together while controlling how outside code can change important values.


```python
class Car:
    def __init__(self, speed, color):
        self._speed = speed
        self.color = color

    def get_speed(self):
        return self._speed

    def set_speed(self, new_speed):
        if new_speed < 0:
            raise ValueError("Speed cannot be negative.")
        self._speed = new_speed
```

```python
car = Car(80, "blue")
print(car.get_speed())
car.set_speed(95)
print(car.get_speed())
```

## Why use controlled access?

If every part of a program could change important data freely, bugs would become much harder to track. Encapsulation gives us one safe place to validate changes.


```python
try:
    car.set_speed(-10)
except ValueError as error:
    print(error)
```

## Naming convention

A single underscore, like `_speed`, is a Python convention that says: “treat this as an internal detail.” It is not absolute security, but it helps communicate intent.
