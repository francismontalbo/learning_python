# Abstraction

Abstraction lets us describe what a class must do without deciding every detail immediately.


## Learning goals

You will learn to:
- define abstract base classes
- require child classes to implement specific methods
- connect abstraction to polymorphism and inheritance


```python
from abc import ABC, abstractmethod


class Notification(ABC):
    @abstractmethod
    def send(self, message):
        """Send a message using a concrete notification channel."""
```

## A concrete implementation


```python
class EmailNotification(Notification):
    def send(self, message):
        print(f"Email sent: {message}")

email_notification = EmailNotification()
email_notification.send("Your project build finished successfully.")
```

## Another implementation


```python
class SMSNotification(Notification):
    def send(self, message):
        print(f"SMS sent: {message}")

for service in [EmailNotification(), SMSNotification()]:
    service.send("Daily reminder: keep practicing Python.")
```

## Connection to polymorphism

Abstraction defines the required behavior. Polymorphism lets different child classes fulfill that behavior in their own way.


```python
def send_all(notifiers, message):
    for notifier in notifiers:
        notifier.send(message)

send_all([EmailNotification(), SMSNotification()], "System maintenance at 8 PM.")
```
