# Inheritance Basics

Inheritance lets one class reuse and extend another class. It is helpful when classes share common behavior.


## A parent class

The parent class contains attributes and methods that are shared by related classes.


```python
class Person:
    def __init__(self, name, role):
        self.name = name
        self.role = role

    def describe(self):
        print(f"{self.name} works as a {self.role}.")
```

```python
class Instructor(Person):
    def __init__(self, name, topic):
        super().__init__(name, "instructor")
        self.topic = topic

    def teach(self):
        print(f"{self.name} is teaching {self.topic}.")
```

```python
class Learner(Person):
    def __init__(self, name, current_topic):
        super().__init__(name, "learner")
        self.current_topic = current_topic

    def study(self):
        print(f"{self.name} is studying {self.current_topic}.")
```

```python
instructor = Instructor("River", "Python basics")
learner = Learner("Avery", "functions")

instructor.describe()
instructor.teach()
learner.describe()
learner.study()
```

## Why `super()` matters

`super()` lets a child class reuse the parent class initialization so we do not repeat ourselves.


```python
print(isinstance(instructor, Instructor))
print(isinstance(instructor, Person))
```
