# Conditions in Python

Programs become interesting when they can make decisions based on data.


## Learning goals

By the end of this notebook, you should be able to:
- compare values with condition operators
- use `if`, `elif`, and `else` blocks
- combine conditions with variables and strings from earlier lessons


## A first condition


```python
temperature = 24

if temperature > 30:
    print("It is a hot day.")
else:
    print("It is not a hot day.")
```

## Equality and inequality


```python
completed_exercises = 10
required_exercises = 10

if completed_exercises == required_exercises:
    print("You met the goal.")
else:
    print("You still have work to do.")
```

```python
project_status = "draft"

if project_status != "published":
    print("The project is still in progress.")
```

## Multiple branches with `elif`


```python
quiz_score = 87

if quiz_score >= 90:
    print("Excellent work.")
elif quiz_score >= 75:
    print("Good progress. Keep practicing.")
else:
    print("Review the lesson and try again.")
```

## Conditions with strings


```python
command = "save"

if command == "run":
    print("Running the script...")
elif command == "save":
    print("Saving your work...")
else:
    print("Unknown command.")
```

## Reusing earlier topics

We can combine conditions with lists, lengths, and f-strings.


```python
pending_tasks = ["review loops", "practice functions"]

if len(pending_tasks) > 0:
    print(f"You still have {len(pending_tasks)} tasks to complete.")
else:
    print("You are done for the day.")
```

## Nested decision-making

You do not always need nested conditions, but it is helpful to recognize them.


```python
study_hours = 3
quiz_score = 92

if study_hours >= 2:
    if quiz_score >= 90:
        print("Strong preparation and strong result.")
    else:
        print("Good effort. Review a few weak spots.")
else:
    print("Try scheduling more practice time.")
```

## Readability tip

Try to keep each condition understandable at a glance. If the logic becomes too tangled, it may belong in a function later.
