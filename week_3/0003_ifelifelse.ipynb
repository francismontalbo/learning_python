# Conditions in Python

Programs become interesting when they can make decisions. In this notebook, we will use `if`, `elif`, and `else` to guide program flow.


## The big idea

A condition asks a question that is either `True` or `False`. Python then chooses which block of code to run.


```python
temperature = 24

if temperature > 30:
    print("It is a hot day.")
else:
    print("It is not a hot day.")
```

## Comparing values

Common comparison operators include `==`, `!=`, `>`, `<`, `>=`, and `<=`. Read each one like plain English.


```python
completed_exercises = 10
required_exercises = 10

if completed_exercises == required_exercises:
    print("You met the goal.")
else:
    print("You still have work to do.")
```

```python
quiz_score = 87

if quiz_score >= 90:
    print("Excellent work.")
elif quiz_score >= 75:
    print("Good progress. Keep practicing.")
else:
    print("Review the lesson and try again.")
```

## Multiple branches

Use `elif` when you have more than two meaningful outcomes.


```python
hour = 14

if hour < 12:
    print("Good morning")
elif hour < 18:
    print("Good afternoon")
else:
    print("Good evening")
```

## Conditions with strings

Conditions also work with text values.


```python
command = "save"

if command == "run":
    print("Running the script...")
elif command == "save":
    print("Saving your work...")
else:
    print("Unknown command.")
```

## Professional tip

Write conditions so that another beginner could understand them without extra explanation. Clear code is a sign of maturity, not simplicity.
