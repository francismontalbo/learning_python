# Truthy and Falsy Values

Python often lets us treat values as true or false without writing an explicit comparison.


## Learning goals

You will learn to:
- recognize common truthy and falsy values
- use concise condition checks
- connect Boolean thinking to lists, strings, and dictionaries from earlier lessons


## Common falsy values


```python
values_to_check = [0, 0.0, "", [], {}, None]

for value in values_to_check:
    if value:
        print(value, "is truthy")
    else:
        print(repr(value), "is falsy")
```

## Common truthy values


```python
examples = [42, "Python", [1, 2, 3], {"topic": "loops"}]

for value in examples:
    if value:
        print(repr(value), "is truthy")
```

## Why this matters

Truthy and falsy checks make code shorter, especially with strings and collections.


```python
pending_tasks = ["read notes", "solve exercise 2"]

if pending_tasks:
    print("You have tasks to finish.")
else:
    print("Your task list is empty.")
```

```python
user_input = ""

if not user_input:
    print("Please type something before continuing.")
```

## Dictionaries and truthiness


```python
profile = {}

if profile:
    print("Profile is ready.")
else:
    print("Profile is still empty.")
```

## Combining with earlier topics

Truthiness is useful inside loops and functions, not just one-time checks.


```python
messages = ["Welcome", "Keep going", ""]

for message in messages:
    if message:
        print(f"Display message: {message}")
    else:
        print("Skip the empty message.")
```

## Readability reminder

Truthy and falsy values are powerful, but do not let brevity hide your meaning.
