# Truthy and Falsy Values

In Python, many values can be treated as `True` or `False` even when they are not literally the Boolean values `True` and `False`. Learning this early will make your code more natural and more readable.


## Falsy values you should know

These commonly evaluate to `False`:
- `0`
- `0.0`
- `""` (an empty string)
- `[]` (an empty list)
- `{}` (an empty dictionary)
- `None`


```python
values_to_check = [0, "", [], {}, None]

for value in values_to_check:
    if value:
        print(value, "is truthy")
    else:
        print(repr(value), "is falsy")
```

## Truthy values

Most non-empty or non-zero values are truthy.


```python
examples = [42, "Python", [1, 2, 3], {"topic": "loops"}]

for value in examples:
    if value:
        print(repr(value), "is truthy")
```

## Why this matters

Truthy and falsy values help us write short, expressive checks.


```python
pending_tasks = ["read notes", "solve exercise 2"]

if pending_tasks:
    print("You have tasks to finish.")
else:
    print("Your task list is empty.")
```

```python
user_input = ""

if not user_input:
    print("Please type something before continuing.")
```

## Readability first

Using truthy and falsy values is powerful, but do not be mysterious. If a condition feels unclear, rewrite it in a more explicit way.
