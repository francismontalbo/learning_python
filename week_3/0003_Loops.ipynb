# Loops in Python

Loops help programs repeat useful work. In this notebook, we will start with `for` loops, then move to `while` loops.


## `for` loops

A `for` loop is a good choice when you want to visit each item in a collection.


```python
students = ["Avery", "Jordan", "Riley", "Sam"]

for student in students:
    print(student)
```

## Doing more inside the loop

The loop variable changes on each pass, so we can build customized output.


```python
for student in students:
    print(f"Welcome, {student}. Keep going.")
```

## Using `range()`

`range()` is helpful when you want to repeat something a specific number of times.


```python
for attempt in range(1, 6):
    print(f"Practice attempt #{attempt}")
```

```python
scores = [78, 85, 91, 88]

for index in range(len(scores)):
    print(f"Score {index}: {scores[index]}")
```

## `while` loops

A `while` loop continues as long as a condition remains true. This is useful when the number of repetitions is not known in advance.


```python
countdown = 3

while countdown > 0:
    print(countdown)
    countdown -= 1

print("Start!")
```

## Loop control

Be careful with `while` loops. If the condition never becomes false, the loop will run forever.


```python
energy = 5

while energy > 0:
    print(f"Working... energy level: {energy}")
    energy -= 1
```
