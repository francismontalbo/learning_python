# Loops in Python

Loops let us repeat useful work without copying the same code again and again.


## Learning goals

You will practice:
- `for` loops over collections
- `range()` for repeated counting
- `while` loops for condition-based repetition
- combining loops with strings, lists, and conditions from earlier notebooks


## A basic `for` loop


```python
students = ["Avery", "Jordan", "Riley", "Sam"]

for student in students:
    print(student)
```

## Using f-strings inside loops


```python
for student in students:
    print(f"Welcome, {student}. Keep going.")
```

## Counting with `range()`


```python
for attempt in range(1, 6):
    print(f"Practice attempt #{attempt}")
```

## Looping through indexes


```python
scores = [78, 85, 91, 88]

for index in range(len(scores)):
    print(f"Score {index}: {scores[index]}")
```

## A loop with conditions

Earlier topics still matter here.


```python
for score in scores:
    if score >= 90:
        print(f"{score} is an excellent result.")
    else:
        print(f"{score} shows progress.")
```

## Building a total with a loop


```python
total_score = 0
for score in scores:
    total_score += score

print("Total score:", total_score)
print("Average score:", total_score / len(scores))
```

## `while` loops

Use a `while` loop when repetition depends on a changing condition.


```python
countdown = 3

while countdown > 0:
    print(countdown)
    countdown -= 1

print("Start!")
```

## Another `while` example


energy = 5

while energy > 0:
    print(f"Working... energy level: {energy}")
    energy -= 1


## Avoiding infinite loops

Always make sure the loop condition can eventually become false.


```python
password_attempts = 0
while password_attempts < 3:
    password_attempts += 1
    print(f"Attempt number {password_attempts}")
```
